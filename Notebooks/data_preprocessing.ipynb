## Data Transformation

```python
df['Gender'] = df['Gender'].map({
    0: "Male",
    1: "Female"
})

df['Internet'] = df['Internet'].map({
    0: "No",
    1: "Yes"
})

df['Extracurricular'] = df['Extracurricular'].map({
    0: "No",
    1: "Yes"
})

df['EduTech'] = df['EduTech'].map({
    0: "Low",
    1: "High"
})

df['Motivation'] = df['Motivation'].map({
    0: "Low",
    1: "Medium",
    2: "High"
})

df['StressLevel'] = df['StressLevel'].map({
    0: "Low",
    1: "Medium",
    2: "High"
})

df['LearningStyle'] = df['LearningStyle'].map({
    0: "Visual",
    1: "Auditory",
    2: "Reading/Writing",
    3: "Kinesthetic"
})

df['FinalGrade'] = df['FinalGrade'].map({
    0: "A",
    1: "B",
    2: "C",
    3: "D"
})
```

```python
print(df.head())
```

## Feature Engineering

```python
df['PerformanceIndex'] = (
    df['ExamScore'] * 0.5 +
    df['Attendance'] * 0.3 +
    df['AssignmentCompletion'] * 0.2
)

df["EngagementScore"] = (
    df["OnlineCourses"] * 0.6 +
    df["Discussions"] * 0.4
)
```

```python
print(df[["PerformanceIndex", "EngagementScore"]].head())
```

## Risk Categorization

```python
def categorize_risk(score):
    if score >= 80:
        return 'Low Risk'
    if score >= 60:
        return 'Medium Risk'
    else:
        return 'High Risk'
    
df['RiskCategory'] = df['PerformanceIndex'].apply(categorize_risk)
```

```python
print(df[['PerformanceIndex', 'RiskCategory']].head())
```

## MySQL Data Loading

```python
from sqlalchemy import create_engine

engine = create_engine(
    "mysql+pymysql://root:root123@localhost:3306/educational_bi_framework"
)

engine.connect()
```

```python
from sqlalchemy import text

with engine.connect() as conn:
    conn.execute(text("SET FOREIGN_KEY_CHECKS = 0"))
    conn.execute(text("TRUNCATE TABLE fact_performance"))
    conn.execute(text("TRUNCATE TABLE dim_students"))
    conn.execute(text("TRUNCATE TABLE dim_behavior"))
    conn.execute(text("TRUNCATE TABLE dim_environment"))
    conn.execute(text("SET FOREIGN_KEY_CHECKS = 1"))
    conn.commit()

print("All tables truncated successfully")
```

```python
students_df = df[[
    "Age",
    "Gender",
    "LearningStyle"
]].copy()

students_df.columns = [
    'age',
    'gender',
    'learning_style'
]

students_df.insert(0, "student_id", range(1, len(students_df) + 1))

students_df.head()
```
